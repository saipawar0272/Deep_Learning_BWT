```python
# to remove missing valyes
df= df.dropna()
```

Transform
Text Cleaning
Remove Punctuation
Remove stopwords
Stemming



```python
import nltk
import re
from nltk.corpus import stopwords  #set of text/values
from nltk.stem import PorterStemmer
ps = PorterStemmer()

corpus = []
for i in range(len(df)):
    rp = re.sub('[^a-zA-Z]', " ", df['Text'][i])
    rp = rp.lower()
    rp = rp.split()
    rp = [ps.stem(word) for word in rp if not word in stopwords.words('english')]
    rp = " ".join(rp)
    corpus.append(rp)
```

```python
corpus 
```

Vectorization
for input variable : count_vectorization
for output variable : label-encoding



```python
corpora                    corpus

sentence                    : mutliple words
parapgraph                   : multiple semtence
lesson/chapter             : multiple paragraph
```

```python
#count vectorization

from sklearn.feature_extraction.text import CountVectorizer  #BOW- Bag of words
cv = CountVectorizer(max_features=None)
x = cv.fit_transform(corpus).toarray()
x
```

```python
x.shape
```

```python
#Label encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Labels'] = le.fit_transform(df[['Labels']])
df['Labels']
```

```python
y = df['Labels']
```

```python
y
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

train the machine wirth ML algorithm using train data
Multicategory Naive-Bayes


```python
from sklearn.naive_bayes import MultinomialNB    #import ML algorithm
model = MultinomialNB()  #save as model name
model.fit(x_train, y_train)
```

#evaluate your model
train accuracy
test accuracy
cross validation


```python
ypred_train = model.predict(x_train)
from sklearn.metrics import accuracy_score
accuracy_score(y_train, ypred_train)
```

```python
ypred_test = model.predict(x_test)
accuracy_score(y_test, ypred_test)
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(model, x,y,cv=5).mean()
```
