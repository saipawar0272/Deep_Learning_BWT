Text Cleaning
1. Remove Punctuation
2. Remove Stopwords
3. Stemming/ Lemmatization

```python
import pandas as pd
df = pd.DataFrame({"in":["He is good boy", "She is good girl", "boy & girl are good"], "op":["P", "p", "N"]})
```

```python
df
```

```python
corpus = []
import re
for i in range(len(df)):
    rp = re.sub('[^a-zA-Z]', " ",df["in"][i])
    rp = rp.lower()
    rp = rp.split()
    rp = [ps.stem(word) for word in rp if not word in stopwords.words('english')]
    rp = " ".join(rp)
    corpus.append(rp)

pd.DataFrame({"in":corpus})
```

Vectorization--> Cout Vectorizer (Bag of Words) --> it counts the reition no.of wordds in the given sentece


```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
pd.DataFrame(cv.fit_transform(corpus).toarray(),columns=cv.get_feature_names_out())
```
