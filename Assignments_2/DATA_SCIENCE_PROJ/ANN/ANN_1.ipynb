```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from sklearn.metrics import mean_squared_error

# Design of Neural Network
nn3 = Sequential()

# Input layer
nn3.add(Input(shape=(4,)))

#  applyting 3 Hidden layers

nn3.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

nn3.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

nn3.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

# Output layer
nn3.add(Dense(units=1,
              activation='linear',
              kernel_initializer='he_uniform'))

# Compile
nn3.compile(optimizer='adam',
            loss='mean_squared_error')

# Train
nn3.fit(x_train,
        y_train,
        batch_size=32,
        epochs=50)

# Predictions
ypred_train = nn3.predict(x_train)
ypred_test = nn3.predict(x_test)

# Train Evaluation
train_mse = mean_squared_error(y_train, ypred_train)
train_rmse = np.sqrt(train_mse)
print("Train RMSE:", train_rmse)

# Test Evaluation
test_mse = mean_squared_error(y_test, ypred_test)
test_rmse = np.sqrt(test_mse)
print("Test RMSE:", test_rmse)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from sklearn.metrics import mean_squared_error

# Design of Neural Network
nn5 = Sequential()

# Input layer
nn5.add(Input(shape=(4,)))

# Hidden layers
nn5.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

nn5.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))
nn5.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

nn5.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

nn5.add(Dense(units=8,
              activation='relu',
              kernel_initializer='uniform'))

# Output layer
nn5.add(Dense(units=1,
              activation='linear',
              kernel_initializer='he_uniform'))


# Compile
nn5.compile(optimizer='adam',
            loss='mean_squared_error')


# Train
nn5.fit(x_train,
        y_train,
        batch_size=32,
        epochs=50)

# Predictions
ypred_train = nn5.predict(x_train)
ypred_test = nn5.predict(x_test)

# Train Evaluation
train_mse = mean_squared_error(y_train, ypred_train)
train_rmse = np.sqrt(train_mse)
print("Train RMSE:", train_rmse)

# Test Evaluation
test_mse = mean_squared_error(y_test, ypred_test)
test_rmse = np.sqrt(test_mse)
print("Test RMSE:", test_rmse)
```
