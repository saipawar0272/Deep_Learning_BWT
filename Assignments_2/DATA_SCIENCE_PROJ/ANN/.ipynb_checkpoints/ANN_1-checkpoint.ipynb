```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2, random_state=42)
```

```python
import tensorflow as tf
import keras
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
!pip install keras
```

#  First  applying Perceptron model --no neurons or layers only i/p and o/p

```python
#initialization of neural network

p_model = Sequential()  #order of excecution

#add layer
p_model.add(Dense(input_dim=4, units=1, activation='relu', kernel_initializer="uniform"))

#compile - calculate

p_model.compile(optimizer='adam', loss='mean_squared_error') #applying gradient descent algo
```

```python
#fit the train data

p_model.fit(x_train, y_train, batch_size=32, epochs=120)
```

```python
#predict on train data
ypred_train = p_model.predict(x_train)

## evaluate the train data
from sklearn.metrics import mean_squared_error
train_mse = mean_squared_error(y_train, ypred_train)
train_rmse = np.sqrt(train_mse)
print("train_RMSE:",train_rmse)
```

```python
#Predicy on test data

ypred_test = p_model.predict(x_test)

##evluate on test data

test_mse = mean_squared_error(y_test, ypred_test)
test_rmse = np.sqrt(test_mse)
print("test RMSE:", test_rmse)
```
