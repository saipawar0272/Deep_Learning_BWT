```python
inp = df.drop(columns='is_arrested')
inp
```

```python
x = pd.get_dummies(inp,drop_first=True)
y = df['is_arrested']
```

```python
df.info()
```

```python
x.shape
```

```python
#splitting the dataswet into training set and test set
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x,y, test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train) # fit always applied on train data not on test data
x_test = sc.transform(x_test)
```

Modelliing

#Design of neural network

#Design
#Fit
#Predict
#Evaluate


```python
from keras.models import Sequential
from keras.layers import Dense

#design of neural perceptron model
p_model= Sequential()
p_model.add(Dense(input_dim=11, units=1, kernel_initializer='uniform', activation='sigmoid'))
p_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

#training the data

p_model.fit(x_train, y_train, epochs=100,batch_size=32)
```

```python
#checking accuracy

y_pred_train = p_model.predict(x_train)
y_pred_train = (y_pred_train >=0.5)

y_pred_test = p_model.predict(x_test)
y_pred_test = (y_pred_test >= 0.5)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
print("accuracy_score:", accuracy_score(y_train, y_pred_train))
print("test_accuracy:", accuracy_score(y_test, y_pred_test))
```

```python
confusion_matrix(y_test, y_pred_test)
```

```python
from keras.models import Sequential
from keras.layers import Dense

#design of neural perceptron model
p_model= Sequential()
p_model.add(Dense(input_dim=11, units=1, kernel_initializer='uniform', activation='sigmoid'))
p_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

#training the data

p_model.fit(x_train, y_train, epochs=100,batch_size=32)
```

applying one hidden layer to the model

```python
n1 = Sequential()
n1.add(Dense(input_dim=11, units=22, kernel_initializer='uniform',activation='relu'))
n1.add(Dense(input_dim=11, units=22, kernel_initializer='uniform',activation='relu'))

n1.add(Dense(units=1, kernel_initializer='uniform', activation='sigmoid'))

n1.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

#training the data

n1.fit(x_train, y_train, epochs=100, batch_size=32)
    
    
```

```python
y_pred_train = n1.predict(x_train)
y_pred_train = (y_pred_train >=0.5)

y_pred_test = n1.predict(x_test)
y_pred_test = (y_pred_test >= 0.5)

```

```python
print("train_accuracy:", accuracy_score(y_train, y_pred_train))
print("test_accuracy:", accuracy_score(y_test, y_pred_test))
```

```python
confusion_matrix(y_test, y_pred_test)
```
